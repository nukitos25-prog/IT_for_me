## Решение задачи

```python
import pandas as pd

# Загрузка данных
df = pd.read_csv('p7_data_13.csv')

# Очистка чисел от $ и запятых
df['salary'] = df['salary'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False).astype(float)
df['Tax'] = df['Tax'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False).astype(float)

# 1. Максимальная и минимальная з/п
print("1. Максимальная и минимальная зарплата:")
print(df.loc[df['salary'].idxmax(), ['LastName', 'FirstName', 'salary']])
print(df.loc[df['salary'].idxmin(), ['LastName', 'FirstName', 'salary']])

# 2. Средняя з/п без указанного навыка
avg_no_skill = df[df['Skill'].isna()]['salary'].mean()
print(f"\n2. Средняя з/п без навыка: {avg_no_skill:.2f}")

# 3. Количество сотрудников без e-mail
no_email = df['email'].isna().sum()
print(f"\n3. Сотрудников без e-mail: {no_email}")

# 4. Суммарный налог
total_tax = df['Tax'].sum()
print(f"\n4. Суммарный налог: {total_tax:.2f}")
```

14. В чем разница между словарными методами pop() и popitem()
применительно к словарям?

pop(key) удаляет элемент с указанным ключом и возвращает его значение.

Если ключа нет, можно вернуть значение по умолчанию или получить KeyError.

popitem() удаляет и возвращает последнюю добавленную пару в виде кортежа (key, value).

В Python 3.7+ порядок вставки сохраняется. Если словарь пуст, возникает KeyError.
