## Задание

В файле pr5_1.txt содержится последовательность целых чисел.
Элементы последовательности могут принимать целые значения от
0 до 10 000 включительно. Рассматривается множество элементов
последовательности, которые оканчиваются на 5 или 7 и при этом
не делятся ни на 9, ни на 11. Найдите количество таких чисел и
сумму минимального и максимального из них.

## Решение

```python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract5/pr5_1.txt
```

```python
with open('pr5_1.txt') as f:
    n = list(map(int, f.read().split()))

z = [x for x in n if x % 9 != 0 and x % 11 != 0 and x % 10 in (5, 7)]
L = len(z)
S = min(z) + max(z)
print(L, S)
```
