## Задание

Напишите функцию, которая печатает таблицу значений функции
$$𝑦 = 𝑥^2 + 3 * 𝑥^3 для 𝑥 ∈ [0, 10].$$ Таблица печатает значения только для целых значений 𝑥 с шагом 1


## Решение

```python
for x in range(0, 11):
    y = x**2 + 3 * x**3
    print(y)
```

# Контрольные вопросы

1. Чем отличается функция с возвратом значения от функций без возврата значения?

Функция с возвратом значения передаёт результат своей работы обратно в точку вызова с помощью оператора return. Этот результат можно присвоить переменной или использовать в выражении. Функция без возврата значения (иногда называемая процедурой) выполняет некоторые действия, но не возвращает явного результата (в Python такая функция возвращает None).

2. Что такое библиотечная функция?

Это функция, которая уже написана и входит в состав стандартной библиотеки языка программирования или сторонней библиотеки. Её можно вызывать в своей программе, не реализуя заново.

3. Почему библиотечные функции похожи на «чёрные ящики»?

Потому что программисту известен их интерфейс (имя, параметры, возвращаемое значение, назначение), но не нужно знать, как именно они устроены внутри. Мы используем функцию, не вникая в детали её реализации.

4. Что делает приведённая ниже инструкция?

x = random.randint(1, 100)

Присваивает переменной x случайное целое число в диапазоне от 1 до 100 включительно. Требуется предварительный импорт модуля random.

5. Что делает приведённая ниже инструкция?

print(random.randint(1, 20))

Выводит на экран случайное целое число от 1 до 20 включительно. (В записи вопроса опечатка: вместо { должна быть (, и лишняя закрывающая скобка.)

6. Для чего используется следующий вызов библиотечной функции?

random.seed(20)

Для инициализации генератора псевдослучайных чисел заданным начальным значением (seed). Это позволяет получать воспроизводимую последовательность случайных чисел: при каждом запуске программы с тем же seed последовательность будет одинаковой.

7. Каким образом функции помогают повторно использовать код в программе?

Один раз определив функцию, её можно вызывать многократно в разных местах программы, передавая различные аргументы. Это избавляет от необходимости копировать один и тот же код и упрощает его сопровождение.

8. Назовите и опишите две части определения функции.

Заголовок — начинается с ключевого слова def, затем имя функции, список параметров в круглых скобках и двоеточие.

Тело — блок инструкций с отступом, который выполняется при вызове функции.

9. Как у функции задать параметры по умолчанию?

В заголовке функции после имени параметра поставить знак равенства и указать значение по умолчанию, например:

def func(a, b=5):

Если при вызове аргумент для b не передан, будет использовано значение 5.

10. Что такое рекурсия?

Это приём, когда функция вызывает саму себя. Обязательно должно быть базовое условие (условие выхода), чтобы рекурсия не была бесконечной.
