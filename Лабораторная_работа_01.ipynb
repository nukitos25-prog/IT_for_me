## Задание

Необходимо вычислить значение выражения:
$$Zze=\frac{35}{f}+y*f-\frac{f+y}{4}$$
Для каждой переменной необходимо вывести ее значение и тип.

## Решение

```python
y = float(input())
f = float(input())
if f != 0:
    print("y = ", y, ", type(y) = ", type(y))
    print("f = ", f, ", type(f) = ", type(f))
    Zze =(35 / f) + (y * f) - ((f + y) / 4)
    print('Zze =',Zze)
    print('Type of Zze: ', type(Zze))
else:
    print("Ошибка: f не должно быть равно 0")
```

# Контрольные вопросы

1. Чем отличается интерпретируемый язык программирования от компилируемого?

Компилируемый язык: исходный код полностью переводится компилятором в машинный код (или промежуточный) до запуска. Ошибки выявляются на этапе компиляции. Программа выполняется быстрее.

Интерпретируемый язык: код выполняется интерпретатором построчно во время работы. Ошибки часто обнаруживаются только при выполнении. Скорость ниже, но выше гибкость и простота отладки.
