## Ответы на контрольные вопросы:


1. Опишите назначение, сходства и различия функций чтения из
текстового файла: read( ), readline( ), readlines( ).

Назначение: все три читают данные из текстового файла.

Сходства: возвращают прочитанное, начиная с текущей позиции; после чтения позиция сдвигается вперёд.

Различия:

read() — читает весь файл (или n символов) и возвращает одну строку.

readline() — читает одну строку (до \n включительно); при повторном вызове — следующую.

readlines() — читает все строки и возвращает список строк (каждый элемент — строка с \n на конце).

2. как прочитать многострочный текстовый файл не используя явно
методы read( ), readline( ), readlines( )?

list(f) — получим список строк.

3. Какова задача закрытия файла?

Освободить системный ресурс, сбросить буферы на диск и разблокировать файл для других программ. Надёжнее всего — через with open(...) as f:, который закрывает файл автоматически.

4. Что такое позиция считывания файла? Где первоначально находится
позиция считывания при открытии файла ввода?

Это «курсор» — номер байта/символа, с которого начнётся следующее чтение или запись. При открытии файла в режиме чтения ('r') позиция находится в начале файла (смещение 0).

5. В каком режиме открывается файл, если требуется записать в него
данные, но при этом требуется оставить существующее содержимое файла
нетронутым? В какую часть файла данные записываются при этом?

Режим 'a' (append). Данные записываются в конец файла; существующее содержимое остаётся нетронутым. (Если файла нет — он создаётся.)

6. Дайте краткое объяснение, что такое исключение.

Это объект-событие, сигнализирующий об ошибке или нештатной ситуации во время выполнения программы (например, деление на ноль, отсутствие файла). Возбуждается (raise) автоматически или вручную.

7. Что происходит, если вызвано исключение, и программа его не
обрабатывает инструкцией try/except?

Оно «всплывает» вверх по стеку вызовов; если нигде не перехвачено try/except, программа аварийно завершается с выводом трассировки (traceback) и сообщения об ошибке.

8. Какое исключение программа вызывает, когда она пытается открыть
несуществующий файл?

FileNotFoundError.

9. Какое исключение программа вызывает, когда она применяет функцию
float( ) для конвертации нечислового строкового значения в число?

ValueError.

10. Что происходит, если файл не существует и программа пытается его
открыть в режиме дозаписи?

Ошибки не будет — файл будет создан пустым и открыт для записи в конец. (То же касается режимов 'w', 'a', 'x' создают файл; 'r' — нет.)
