## Задание

Класс «График y=3x+5». Реализовать ввод и вывод полей данных,
вычисление интеграла функции от a до b (вводятся пользователем),
длины отрезка функции от (a, y(a)) до (b, y(b)), а также вывод
информации об объекте.

## Решение

```python
import math

class Graf:
    def __init__(self):
        self.k = 3
        self.b = 5

    def y(self, x):
        return self.k * x + self.b

    def integral(self, a, b):
        return (self.y(a) + self.y(b)) * (b - a) / 2

    def dlina(self, a, b):
        return math.sqrt((b - a)**2 + (self.y(b) - self.y(a))**2)

    def info(self):
        print("График y =", self.k, "x +", self.b)


g = Graf()
g.info()

a = float(input("a = "))
b = float(input("b = "))

print("Точки:", (a, g.y(a)), "и", (b, g.y(b)))
print("Интеграл:", g.integral(a, b))
print("Длина отрезка:", g.dlina(a, b))
```
