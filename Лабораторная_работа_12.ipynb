## Решение

### Подключение библиотек и загрузка данных

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
%matplotlib inline

# Загрузка набора данных tips (скачивается из интернета)
tips = sns.load_dataset('tips')
tips.head(10)
```

```python
tips.info()
```

### Одиночные признаки: количественные

```python
# Гистограмма распределения суммы счёта
# bins=20 задаёт количество интервалов
tips['total_bill'].hist(bins=20)
plt.xlabel('Total bill, $')
plt.ylabel('Frequency')
plt.title('Distribution of total bill')
plt.show()
```

```python
# Ящик с усами для размера чаевых
# Горизонтальное расположение (orient='h') удобно для одного признака
sns.boxplot(x=tips['tip'], orient='h')
plt.xlabel('Tip, $')
plt.title('Boxplot of tips')
plt.show()
```

### Дополнительно: гистограмма для дискретного признака size

```python
# Гистограмма для количества человек за столом (дискретный признак)
tips['size'].hist(bins=6, edgecolor='black')
plt.xlabel('Table size (number of people)')
plt.ylabel('Frequency')
plt.title('Distribution of table size')
plt.show()
```

11. Поясните методы установки размера графика.

Основные способы:

Через matplotlib:

plt.figure(figsize=(10, 6))

Через subplots:

fig, ax = plt.subplots(figsize=(10, 6))

Через seaborn:

sns.set(rc={'figure.figsize': (10, 6)})

Через pandas:

data['col'].hist(figsize=(10, 6))

Изменить размер уже созданной фигуры:

fig.set_size_inches(10, 6)

12. Опишите функции, с использованием которых можно решить
следующие задачи: установить шрифт для осей графика, установить подписи осей
графика, установить текст легенды графика.

Установить шрифт для осей графика

Можно задать размер шрифта подписей осей и делений:

plt.xlabel('X axis', fontsize=14)

plt.ylabel('Y axis', fontsize=14)

plt.xticks(fontsize=12)

plt.yticks(fontsize=12)

Или через fontdict:

plt.xlabel('X axis', fontdict={'fontsize': 14, 'fontweight': 'bold'})

Для глобальной настройки:

plt.rcParams['font.size'] = 12

plt.rcParams['font.family'] = 'DejaVu Sans'

В объектном интерфейсе:

ax.set_xlabel('X axis', fontsize=14)

ax.set_ylabel('Y axis', fontsize=14)

ax.tick_params(axis='both', labelsize=12)

Установить подписи осей графика

plt.xlabel('Название оси X')

plt.ylabel('Название оси Y')

plt.title('Заголовок графика')

Или:

ax.set_xlabel('Название оси X')

ax.set_ylabel('Название оси Y')

ax.set_title('Заголовок графика')

Установить текст легенды графика

Текст легенды задаётся через параметр label при построении графика:

plt.plot(x, y1, label='Первый ряд')

plt.plot(x, y2, label='Второй ряд')

plt.legend(title='Обозначения', loc='best', fontsize=12)

В объектном интерфейсе:

ax.plot(x, y1, label='Первый ряд')

ax.legend(title='Обозначения', loc='upper right', fontsize=12)
