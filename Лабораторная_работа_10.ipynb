## Решение

Файл `pr7_13.txt` расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

```python
import pandas as pd

# 1. Загрузка
df = pd.read_csv('p7_data_13.csv')

# 2. Фильтрация: зарплата > 5000, налог < 400
filtered = df[
    (df['salary'].apply(lambda x: float(x.replace('$', '')) > 5000)) &
    (df['Tax'].apply(lambda x: float(x.replace('$', '')) < 400))
]

# 3. Сортировка: по числовому значению зарплаты (убывание)
sorted_df = filtered.sort_values(
    by='salary',
    key=lambda col: col.str.replace('$', '').astype(float),
    ascending=False
)

# 4. Отображение: создание столбца "зарплата в евро" и категории
sorted_df['salary_eur'] = sorted_df['salary'].map(
    lambda x: round(float(x.replace('$', '')) * 0.92, 2)
)
sorted_df['category'] = sorted_df['salary'].apply(
    lambda x: 'High' if float(x.replace('$', '')) > 7000 else 'Medium'
)

# 5. Вывод результата
print(sorted_df[['LastName', 'FirstName', 'salary', 'salary_eur', 'category']])
```

1. Что такое анонимная функция?

Анонимная функция — это функция без имени, которая создаётся «на месте» и обычно используется один раз. В Python для этого служит lambda-выражение.

lambda параметры: выражение

Особенности:

возвращает результат выражения автоматически, без return;

внутри можно записать только одно выражение, а не блок инструкций;

может принимать ноль, один или несколько аргументов.

Лямбда-выражения нужны для компактной передачи простой логики в другие функции: map(), filter(), sorted(), sort(), apply() в pandas и т.п.

Возведение в квадрат через map

nums = [1, 2, 3, 4]

squares = list(map(lambda x: x ** 2, nums))

print(squares)  # [1, 4, 9, 16]

Сортировка по последнему символу строки

words = ['кот', 'слон', 'мышь']

print(sorted(words, key=lambda w: w[-1]))

Сумма двух чисел

add = lambda a, b: a + b

print(add(2, 3))  # 5
