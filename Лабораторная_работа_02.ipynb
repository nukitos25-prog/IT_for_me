## Задание

Необходимо вычислить значение выражения:
$$Z=1-\frac{sin(X^2)+Y}{2}+\frac{cos(Y^3)+X}{3}-\frac{sin(X^4)+Y}{4}+\frac{cos(Y^5)+X}{5}-...$$
Для каждой переменной необходимо вывести ее значение и тип.

## Решение

```python
Z = 1
N = int(input())
X = float(input())
Y = float(input())
for i in range(2, N+1):
    chisl1 = (X)**(i)+Y
    chisl2 = (Y)**(i)+X
    znam = i
    if i % 2 == 1:
        Z += chisl2 / znam
    else:
        Z -= chisl1 / znam
print(Z)
```
